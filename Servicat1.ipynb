```python
# --- PASO 2: Cargar el archivo completo ---
import pandas as pd

ruta = '/content/drive/MyDrive/base_datos_servicat_pbi 1 (4).xlsx'
df = pd.read_excel(ruta, sheet_name=['ventas','productos','clientes','vendedores','sucursales','metas_mensuales'])
```

```python
# Fase 2: Limpiar los datos

# --- PASO 3: Sacar cada tabla del diccionario ---
ventas = df['ventas']
productos = df['productos']
clientes = df['clientes']
vendedores = df['vendedores']
sucursales = df['sucursales']
metas = df['metas_mensuales']
```

```python
# --- PASO 4: Resumen general de cada tabla (filas, nulos, tipos de dato) ---
print("VENTAS")
ventas.info()

print("\nPRODUCTOS")
productos.info()

print("\nCLIENTES")
clientes.info()

print("\nVENDEDORES")
vendedores.info()

print("\nSUCURSALES")
sucursales.info()

print("\nMETAS")
metas.info()
```

```python
# --- PASO 6: Revisar valores nulos (celdas vacías) ---
print("Nulos en VENTAS")
print(ventas.isnull().sum())

print("Nulos en SUCURSALES")
print(sucursales.isnull().sum())

print("Nulos en METAS")
print(metas.isnull().sum())
```

```python
# --- PASO 7: Revisar duplicados ---
print("Duplicados en VENTAS:", ventas.duplicated().sum())
print("Duplicados en SUCURSALES:", sucursales.duplicated().sum())
print("Duplicados en METAS:", metas.duplicated().sum())
```

```python
# --- PASO 8: Revisar los tipos de dato de cada columna ---
print("\nTipos de dato en VENTAS")
print(ventas.dtypes)

print("\nTipos de dato en SURCUSALES")
print(sucursales.dtypes)

print("\nTipos de dato en METAS")
print(metas.dtypes)
```

```python
# --- PASO 9: Limpiar (borrar nulos y duplicados, si los hubiera) ---
ventas_limpio = ventas.dropna().drop_duplicates()
metas_limpio = metas.dropna().drop_duplicates()
sucursales_limpio = sucursales.dropna().drop_duplicates()

# Confirmar cuántas filas quedaron después de limpiar
print("\nVentas: antes =", len(ventas), "| después =", len(ventas_limpio))
print("Metas: antes =", len(metas), "| después =", len(metas_limpio))
print("Sucursales: antes =", len(sucursales), "| después =", len(sucursales_limpio))
```

```python
# Fase 3: Transformar y preparar los datos

# --- PASO 10: Sacar año, mes y calcular el total de cada venta ---
ventas_limpio['anio'] = ventas_limpio['fecha'].dt.year
ventas_limpio['mes'] = ventas_limpio['fecha'].dt.month
ventas_limpio['total_venta'] = ventas_limpio['cantidad'] * ventas_limpio['precio_unitario']

display(ventas_limpio.head())
```

```python
# --- PASO 11: Sumar el total de ventas por sucursal, año y mes ---
ventas_por_sucursal = ventas_limpio.groupby(['id_sucursal', 'anio', 'mes'])['total_venta'].sum().reset_index()

print(ventas_por_sucursal.head())
```

```python
# --- PASO 12: Unir las ventas reales con las metas ---
comparacion = ventas_por_sucursal.merge(metas_limpio, on=['id_sucursal', 'anio', 'mes'])

print(comparacion.head())
```

```python
## Fase 4: Analizar
```

```python
# --- PASO 13: Calcular el porcentaje de cumplimiento de meta ---
comparacion['cumplimiento_pct'] = (comparacion['total_venta'] / comparacion['meta_ventas']) * 100

print(comparacion.head())
```

```python
# --- PASO 14: Ordenar de peor a mejor cumplimiento ---
peores = comparacion.sort_values('cumplimiento_pct').head(10)

display(peores[['nombre_sucursal', 'mes_nombre', 'total_venta', 'meta_ventas', 'cumplimiento_pct']])
```

```python
## Los peores casos de incumplimiento no se concentran en una sola tienda, aparecen repartidos en distintas tiendas y meses,
## lo que sugiere un problema general de la empresa, no de una sola sucursal y no llegan con la meta establecida.
```

```python
# --- PASO 15: Promedio general y % de cumplimiento de toda la empresa ---
promedio_general = comparacion['cumplimiento_pct'].mean()
print("Promedio general de cumplimiento:", promedio_general)

pct_cumplio = (comparacion['cumplimiento_pct'] >= 100).mean() * 100
print("% de meses que SÍ cumplieron la meta:", pct_cumplio)

pct_no_cumplio = (comparacion['cumplimiento_pct'] < 100).mean() * 100
print("% de meses que NO cumplieron la meta:", pct_no_cumplio)
```

```python
## El 90% de los meses, las tiendas de Servicat no llegan a su meta de ventas y solo el 10% lo hace.
## El cual genera inestabilidad a la empresa. El cumplimiento general es del 60%
```

```python
# --- PASO 16: Traer la ciudad y calcular el promedio de cumplimiento por ciudad ---
comparacion = comparacion.merge(sucursales_limpio[['id_sucursal', 'ciudad']], on='id_sucursal')

promedio_por_ciudad = comparacion.groupby('ciudad')['cumplimiento_pct'].mean().sort_values()
print(promedio_por_ciudad)
```

```python
## El cumplimiento de metas varía significativamente por ciudad: Santo Domingo (44.6%) y Ambato (44.7%) presentan el desempeño más bajo,
## mientras que Cuenca y Cumbayá (81.3% cada una) muestran el mejor cumplimiento relativo. La brecha de casi 37 puntos porcentuales entre
## la ciudad más débil y la más fuerte indica que el problema de incumplimiento no es homogéneo, sino que está concentrado geográficamente.
```

```python
# --- PASO 17: Ventas totales por categoría de producto ---
ventas_categoria = ventas_limpio.merge(productos[['id_producto', 'categoria']], on='id_producto')

resumen_categoria = ventas_categoria.groupby('categoria')['total_venta'].sum().sort_values(ascending=False)
print(resumen_categoria)
```

```python
# --- Extra: ticket promedio por categoría ---
ticket_promedio = ventas_categoria.groupby('categoria')['total_venta'].mean().sort_values(ascending=False)
print(ticket_promedio)
```

```python
## El ticket promedio confirma por qué Muebles genera más dinero que las demás categorías: cada compra de Muebles cuesta en promedio $3,005,
## mientras que una compra de Limpieza cuesta apenas $198.75. Es decir, Muebles no gana por venderse más veces, sino porque cada venta individual
## vale mucho más dinero.
```
